```python
#creating_filter_for_asia
mask=df['origin']=='Asia'
```

```python
#dataset for asia
df_Asia=df[mask]
```

```python
#selecting only vechile_weight and model_year
df_vw_my=df_Asia[['vehicle_weight','model_year']]
```

```python
#selecting_first_Seven_entries
df_seven=df_vw_my.head(7)
```

```python
#converting into array
X=df_seven.to_numpy()
```

```python
#multiplying x transpose with x
XTX=np.matmul(X.T,X)
```

```python
#inverting xtx
XTX_INV=np.linalg.inv(XTX)
```

```python
#creating y
y=np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
#performing_multiplication
```

```python
Z=np.matmul(XTX_INV,X.T)
```

```python
w=np.matmul(Z,y)
```

```python
#sum of elements
s=np.sum(w)
```
